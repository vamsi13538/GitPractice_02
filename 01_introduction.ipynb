# <u>Git & GitHub</u>

# <u>What is Git?</u>
- Git is a distributed version control system (VCS) used to track changes in source code and files during software development. It helps developers manage different versions of a project, collaborate with others, and restore previous versions when needed.

### Key Features:
- Tracks file and code changes.
- Supports branching and merging.
- Enables collaboration among multiple developers.
- Works locally on your computer.
- Maintains a complete history of project changes.

### Example:
- A developer can save code changes using commits and revert to an earlier version if a mistake is made.

# <u>What is GitHub?</u>
- GitHub is a cloud-based platform that hosts Git repositories online. It provides tools for collaboration, code review, project management, and continuous integration.

### Key Features:
- Online storage for Git repositories.
- Team collaboration through pull requests.
- Issue tracking and project management.
- Access control and version history.
- Integration with DevOps and CI/CD tools.

### Example:
- Developers can upload their Git repositories to GitHub and work together on the same project from different locations.

# <u>Git Vs. GitHub</u>

**In short:** Git is the tool used to track code changes, while GitHub is the platform used to store, share, and collaborate on Git repositories.

# <u>What is Version Control System (VCS) ?</u>
- Version Control is a system that helps track and manage changes made to files, documents, or source code over time. It allows multiple people to work on the same project, maintain a history of changes, and restore previous versions when needed.

### Key Features:
- Tracks all changes made to files.
- Maintains a history of revisions.
- Allows multiple users to collaborate.
- Supports branching and merging of work.
- Enables rollback to previous versions if errors occur.

### Benefits:
- Prevents loss of work.
- Improves team collaboration.
- Makes it easy to identify who made specific changes.
- Helps manage different versions of a project efficiently.

**Example:** When developers work on a software project, version control records every code change, allowing them to revert to earlier versions or merge contributions from multiple team members.

**In short:** Version Control is a system used to track, manage, and organize changes to files and code over time.
